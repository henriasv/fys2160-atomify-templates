# A box of gas at constant pressure

The box is heated from T* = 0.6 to 3.0 and cooled back at constant pressure P* = 0.02. For an ideal gas the area is A = N k T / P, a straight line through the origin. Plot the area against the temperature and compare with that line (N = 400).

```python
import glob
import lammps_logfile
import matplotlib.pyplot as plt

logs = sorted(glob.glob("runs/*/log.lammps"))   # every run of this project
log = lammps_logfile.File(logs[-1])             # newest run
print(log.get_keywords())
```

```python
T = log.get("Temp", run_num=-1); A = log.get("Volume", run_num=-1)
plt.plot(T, A, ".", ms=3, label="simulation")
plt.plot([0, T.max()], [0, 400 / 0.02 * T.max()], "--", label="ideal gas  A = N k T / P")
plt.xlabel("T*"); plt.ylabel("area (σ²)"); plt.legend(); plt.show()
```
