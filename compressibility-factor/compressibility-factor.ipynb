Open a completed run and inspect the available data. After each simulation, rerun the load cell, the data/plot cell, and your calculation cell in that order.


```python
import lammps_logfile
import glob

logs = sorted(glob.glob("runs/*/log.lammps"))
if not logs:
    raise RuntimeError("No runs yet: complete a simulation, then rerun this cell.")
print("Runs found:", *logs, sep="\n  ")
selected_log = logs[-1]  # or choose an earlier completed run from the list
print("Selected log:", selected_log)
log = lammps_logfile.File(selected_log)
print("Log keywords:", log.get_keywords())
```

Choose and import data from logfile and plot data.

```python
t = log.get("Time")
T = log.get("Temp")
P = log.get("Press")

import matplotlib.pyplot as plt
plt.plot(t, T)
plt.xlabel('Time')
plt.ylabel('Temp')
plt.title('Temp')
plt.show()
plt.plot(t, P)
plt.xlabel('Time')
plt.ylabel('Pressure')
plt.title('Pressure')
plt.show()
```
