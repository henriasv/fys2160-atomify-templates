Open logfilfe and display which data are available and what the keywords for accessing this data are.

```python
import lammps_logfile
import matplotlib
# Atomify keeps every run of this project in runs/<run-name>/; take the newest.
import glob
log = lammps_logfile.File(sorted(glob.glob("runs/*/log.lammps"))[-1])
print("Log keywords: ", log.get_keywords())
```

Choose and import data from logfile and plot data.

```python
t = log.get("Time")
T = log.get("Temp")
P = log.get("Press")

import matplotlib.pyplot as plt
plt.plot(t, T)
plt.xlabel('Time')
plt.ylabel('Temp')
plt.title('Temp')
plt.show()
plt.plot(t, P)
plt.xlabel('Time')
plt.ylabel('Pressure')
plt.title('Pressure')
plt.show()
```
