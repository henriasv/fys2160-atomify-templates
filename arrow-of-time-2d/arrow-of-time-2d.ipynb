# The arrow of time

All atoms start in the left half. After `trev` steps every velocity is reversed. Plot the number of atoms in the left half against time: does it return to N? Run again with a larger `trev` (edit the `variable trev index ...` line) and compare.

```python
import glob
import lammps_logfile
import matplotlib.pyplot as plt

logs = sorted(glob.glob("runs/*/log.lammps"))   # every run of this project
log = lammps_logfile.File(logs[-1])             # newest run
print(log.get_keywords())
```

```python
t = log.get("Time"); n = log.get("v_nleft")
plt.plot(t, n)
plt.axhline(n[0] / 2, ls=":", color="gray")
plt.xlabel("time t*"); plt.ylabel("atoms in the left half"); plt.show()
```
